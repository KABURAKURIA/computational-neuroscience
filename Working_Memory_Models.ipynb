# Working Memory Models
In this notebook, we explore continuous attractor models of working memory.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

N = 100
tau = 10.0
dt = 1.0
T = 100

w = np.zeros((N, N))
for i in range(N):
    for j in range(N):
        w[i, j] = 2.0 * np.exp(-((i-j)**2)/20.0) - 1.0

r = np.random.rand(N) * 0.1
rates = []

for t in range(int(T/dt)):
    I_ext = np.zeros(N)
    if 10 < t < 20:
        I_ext[40:60] = 5.0
    dr = (-r + np.maximum(0, np.dot(w, r) + I_ext)) / tau
    r += dr * dt
    rates.append(r.copy())

plt.imshow(np.array(rates).T, aspect="auto", origin="lower")
plt.xlabel("Time")
plt.ylabel("Neuron index")
plt.title("Bump Attractor Working Memory")
plt.show()
```
