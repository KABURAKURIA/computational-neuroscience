# Grid Cells and Place Cells
In this notebook, we visualize idealized responses of grid cells and place cells in 2D space.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def place_cell_response(x, y, center_x, center_y, sigma=0.5):
    return np.exp(-((x - center_x)**2 + (y - center_y)**2) / (2 * sigma**2))

def grid_cell_response(x, y, spacing=2.0):
    k1 = np.array([0, 1]) * 4 * np.pi / (spacing * np.sqrt(3))
    k2 = np.array([np.sqrt(3)/2, -0.5]) * 4 * np.pi / (spacing * np.sqrt(3))
    k3 = np.array([-np.sqrt(3)/2, -0.5]) * 4 * np.pi / (spacing * np.sqrt(3))
    
    val = np.cos(k1[0]*x + k1[1]*y) + np.cos(k2[0]*x + k2[1]*y) + np.cos(k3[0]*x + k3[1]*y)
    return (val + 1.5) / 3.0

xx, yy = np.meshgrid(np.linspace(-5, 5, 100), np.linspace(-5, 5, 100))

pc = place_cell_response(xx, yy, 0, 0)
gc = grid_cell_response(xx, yy)

fig, axs = plt.subplots(1, 2, figsize=(12, 5))
axs[0].imshow(pc, extent=(-5, 5, -5, 5), origin='lower', cmap='jet')
axs[0].set_title('Place Cell Response')
axs[1].imshow(gc, extent=(-5, 5, -5, 5), origin='lower', cmap='jet')
axs[1].set_title('Grid Cell Response')
plt.show()
```
