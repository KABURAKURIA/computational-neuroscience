## 1. Dopamine Modulated STDP
Neuromodulation can gate plasticity.

```python

start_scope()
N = 100
tau = 10*ms
eqs = '''
dv/dt = (v0 - v) / tau : volt (unless refractory)
v0 : volt
'''
G = NeuronGroup(N, eqs, threshold='v > -50*mV', reset='v = -70*mV', refractory=2*ms, method='euler')
G.v = -70*mV
G.v0 = -70*mV
```
