# Predictive Coding
Predictive coding posits that the brain is continuously generating models of the world and updating them based on sensory input. Here we implement a simple hierarchical network.


```python
import numpy as np
import matplotlib.pyplot as plt

input_signal = np.array([0.8, 0.4])
weights = np.array([[1.0, 0.5], [0.5, 1.0]])
representation = np.array([0.0, 0.0])

learning_rate = 0.05
epochs = 100

errors = []
rep_history = []

for i in range(epochs):
    prediction = np.dot(weights, representation)
    error = input_signal - prediction
    
    errors.append(np.sum(error**2))
    
    representation += learning_rate * np.dot(weights.T, error)
    rep_history.append(representation.copy())

rep_history = np.array(rep_history)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(errors)
plt.title('Prediction Error over Time')
plt.xlabel('Epoch')
plt.ylabel('Squared Error')

plt.subplot(1, 2, 2)
plt.plot(rep_history[:, 0], label='Neuron 1')
plt.plot(rep_history[:, 1], label='Neuron 2')
plt.title('Internal Representation')
plt.xlabel('Epoch')
plt.legend()

plt.tight_layout()
plt.show()
```
