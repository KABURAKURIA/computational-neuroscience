# Reinforcement Learning in Neuroscience
This notebook covers fundamental concepts of reinforcement learning in the brain, such as the Rescorla-Wagner model and Temporal Difference (TD) learning, which models the activity of midbrain dopamine neurons.

```python
import numpy as np
import matplotlib.pyplot as plt

print('Libraries loaded!')
```

## 1. Rescorla-Wagner Model
The Rescorla-Wagner model describes classical conditioning. The change in the value of a stimulus $V$ is proportional to the prediction error: $\Delta V = \alpha (R - V)$, where $R$ is the reward and $\alpha$ is the learning rate.

```python
# Rescorla-Wagner Simulation
n_trials = 50
alpha = 0.1 # Learning rate
rewards = np.ones(n_trials) # Constant reward of 1
V = np.zeros(n_trials + 1) # Value estimates

for t in range(n_trials):
    prediction_error = rewards[t] - V[t]
    V[t+1] = V[t] + alpha * prediction_error

plt.figure(figsize=(8, 4))
plt.plot(V[:-1], marker='o', label='Estimated Value (V)')
plt.axhline(1.0, color='r', linestyle='--', label='Actual Reward')
plt.xlabel('Trials')
plt.ylabel('Value')
plt.title('Rescorla-Wagner Model: Acquisition')
plt.legend()
plt.show()
```
