# Izhikevich Neuron Model
This notebook demonstrates the Izhikevich model, known for reproducing diverse spiking patterns with biological realism and computational efficiency.

```python
import numpy as np
import matplotlib.pyplot as plt
from brian2 import *
%matplotlib inline
```

## Regular Spiking
Simulating a regular spiking neuron.

```python
start_scope()
a = 0.02
b = 0.2
c = -65
d = 8
eqs = '''
dv/dt = (0.04 * v**2 + 5 * v + 140 - u + I) / ms : 1
du/dt = a * (b * v - u) / ms : 1
I : 1
'''
G = NeuronGroup(1, eqs, threshold='v > 30', reset='v = c; u = u + d', method='euler')
G.v = c
G.u = b * c
G.I = 10
M = StateMonitor(G, 'v', record=True)
run(100*ms)
plt.plot(M.t/ms, M.v[0])
plt.xlabel('Time (ms)')
plt.ylabel('v')
plt.show()
```
