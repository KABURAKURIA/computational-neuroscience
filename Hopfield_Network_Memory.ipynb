## 1. Storing Patterns
We store binary patterns using Hebbian learning.

```python

N = 100
P = 5
patterns = np.random.choice([-1, 1], size=(P, N))
W = np.zeros((N, N))
for p in patterns:
    W += np.outer(p, p)
np.fill_diagonal(W, 0)
W /= N
```
