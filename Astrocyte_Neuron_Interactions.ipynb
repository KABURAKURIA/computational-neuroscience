# Astrocyte-Neuron Interactions
In this notebook, we explore mathematical models of astrocyte-neuron interactions.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

dt = 0.1
T = 100.0
time = np.arange(0, T, dt)
Ca = np.zeros_like(time)
IP3 = np.zeros_like(time)

for i in range(1, len(time)):
    stim = 1.0 if 20 < time[i] < 40 else 0.0
    dIP3 = (stim - IP3[i-1]) * 0.1
    dCa = (IP3[i-1] - Ca[i-1]) * 0.05
    IP3[i] = IP3[i-1] + dIP3 * dt
    Ca[i] = Ca[i-1] + dCa * dt

plt.plot(time, Ca, label="Calcium")
plt.plot(time, IP3, label="IP3")
plt.xlabel("Time (ms)")
plt.ylabel("Concentration")
plt.legend()
plt.title("Astrocyte Calcium Dynamics")
plt.show()
```
