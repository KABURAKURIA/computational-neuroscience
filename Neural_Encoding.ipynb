## 1. Poisson Spike Generation
Neurons often emit spikes that can be approximated as a Poisson process. Given a firing rate $\lambda(t)$, we can generate spike trains.

```python
def generate_poisson_spikes(rate, duration, dt=0.001):
    """
    Generate a Poisson spike train.
    rate: firing rate in Hz
    duration: duration in seconds
    dt: time step in seconds
    """
    n_bins = int(duration / dt)
    prob_of_spike = rate * dt
    spikes = np.random.rand(n_bins) < prob_of_spike
    return spikes

# Generate spikes for 1 second at 50 Hz
rate = 50.0
duration = 1.0
dt = 0.001
spikes = generate_poisson_spikes(rate, duration, dt)
time = np.arange(0, duration, dt)

plt.figure(figsize=(10, 2))
plt.eventplot(np.where(spikes)[0] * dt, color='black')
plt.title(f'Poisson Spike Train (Rate = {rate} Hz)')
plt.xlabel('Time (s)')
plt.ylabel('Neuron')
plt.yticks([])
plt.show()
```
