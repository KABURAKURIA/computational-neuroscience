# Bayesian Inference in the Brain
How the brain might represent and compute with probabilities.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
%matplotlib inline
```

## 1. Cue Integration
Combining visual and auditory cues.

```python

x = np.linspace(-10, 10, 1000)
mu_v, sigma_v = 1.0, 1.5
mu_a, sigma_a = 2.0, 2.5
visual = stats.norm.pdf(x, mu_v, sigma_v)
auditory = stats.norm.pdf(x, mu_a, sigma_a)
w_v = (1/sigma_v**2) / (1/sigma_v**2 + 1/sigma_a**2)
w_a = (1/sigma_a**2) / (1/sigma_v**2 + 1/sigma_a**2)
mu_combined = w_v * mu_v + w_a * mu_a
sigma_combined = np.sqrt(1 / (1/sigma_v**2 + 1/sigma_a**2))
combined = stats.norm.pdf(x, mu_combined, sigma_combined)

plt.plot(x, visual, label='Visual (Likelihood)')
plt.plot(x, auditory, label='Auditory (Prior)')
plt.plot(x, combined, label='Combined (Posterior)', linewidth=2)
plt.legend()
plt.show()
```
