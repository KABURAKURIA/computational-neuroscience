# Sensory Processing: Visual Cortex Models
In this notebook, we explore sensory processing by modeling the receptive fields of simple cells in the primary visual cortex (V1) using Gabor filters.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## 1. Modeling V1 Receptive Fields with Gabor Filters
A Gabor filter is a linear filter used for edge detection, representing the properties of spatial frequency and orientation selectivity of V1 simple cells.

```python
def gabor_patch(size, lambda_, theta, sigma, gamma, psi):
    """Generates a Gabor patch."""
    x, y = np.meshgrid(np.linspace(-size/2, size/2, size), np.linspace(-size/2, size/2, size))
    
    # Rotation
    x_theta = x * np.cos(theta) + y * np.sin(theta)
    y_theta = -x * np.sin(theta) + y * np.cos(theta)
    
    # Gabor function
    envelope = np.exp(-(x_theta**2 + gamma**2 * y_theta**2) / (2 * sigma**2))
    carrier = np.cos(2 * np.pi * x_theta / lambda_ + psi)
    
    return envelope * carrier

# Parameters
size = 100
lambda_ = 10  # Wavelength
theta = np.pi/4  # Orientation (45 degrees)
sigma = 10  # Standard deviation of Gaussian envelope
gamma = 0.5  # Spatial aspect ratio
psi = 0  # Phase offset

gabor = gabor_patch(size, lambda_, theta, sigma, gamma, psi)

plt.imshow(gabor, cmap='gray')
plt.title('Gabor Filter (V1 Receptive Field Model)')
plt.axis('off')
plt.colorbar()
plt.show()
```
